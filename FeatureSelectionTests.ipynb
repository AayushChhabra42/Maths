- Hypothesis Testing: infer about a population based on a sample or correlation between two or more samples
- Null Hypothesis(H0): This hypothesis states that there is no significant difference between sample and population or among different populations
- Alternate Hypothesis(H1): The statement contrary to the null hypothesis comes under the alternate hypothesis
- Critical Value: It is a point on the scale of the test statistic beyond which the null hypothesis is rejected.
- p-value: p-value stands for ‘probability value’; it tells how likely it is that a result occurred by chance alone. Basically, the p-value is used in hypothesis testing to help you support or reject the null hypothesis. The smaller the p-value, the stronger the evidence to reject the null hypothesis.
- Degree of freedom: The degree of freedom is the number of independent variables.

A statistical test is a way to determine whether the random variable is following the null hypothesis or alternate hypothesis.

The procedure for implementing each statistical test will be as follows:

- We calculate the statistic value using the mathematical formula
- We then calculate the critical value using statistic tables
- With the help of critical value, we calculate the p-value
- If p-value> 0.05 we accept the null hypothesis else we reject it

```python
import numpy as np
import pandas as pd
import seaborn as sns
from numpy import sqrt, abs, round
import scipy.stats as stats
from scipy.stats import norm
```

```python
class statCalc:
  def Zstat2(self,u1,u2,mudiff,sd1, sd2, n1, n2):
    '''
    Calculates 2 sample Z statistic
    Note: Do not use Z test for sample size less than 30
    '''
    pooledSE = sqrt(sd1**2/n1 + sd2**2/n2)
    z = ((u1 - u2) - mudiff)/pooledSE
    pval = 2*(1 - norm.cdf(abs(z)))
    return round(z,3), pval

  def Ttest(self, u1, u2, sd1, sd2, n1, n2):
    '''
    A t-test is also used to compare the mean of two given samples like the Z-test.
    However, It is implemented when the sample data size is less than 30.
    It assumes a normal distribution of the sample. It can also be one-sample or two-sample.
    The degree of freedom is calculated by n-1 where n is the number of samples.
    '''
    t_stat, p_val = stats.ttest_ind_ind(u1, u2, equal_var=False) # Assuming unequal variances for a more general case
    return round(t_stat, 3), p_val

  def CorrelationTest(self, x, y, type='Pearson'):
    '''
    Calculates pearson correlation between two variables
    '''
    if type == 'Pearson':
      corr=np.corrcoef(x, y)
    else:
      corr, p_val = None, None # Handle other correlation types if needed
    return round(corr, 3), p_val

  def AnovaTest(self, *args, type='OneWay'):
    '''
    One-way ANOVA is applied when there are three or more independent groups of a variable.
    '''
    if type == 'OneWay':
      f_stat, p_val = stats.f_oneway(*args)
    else:
      f_stat, p_val = None, None # Handle other ANOVA types if needed
    return round(f_stat, 3), p_val

  def ChiSquareTest(self, observed):
    '''
    It is used to determine whether there is a significant association or relationship between the two variables.
    '''
    chi2, p_val, dof, expected = stats.chi2_contingency(observed)
    return round(chi2, 3), p_val, dof, expected
```

Example
One categorical variable(Gender) against a continuous variable(Income)

- Null Hypothesis: There is no significant difference between the mean Income of males and females.

- Alternate Hypothesis: There is a significant difference between the mean Income of males and females.



```python
df=pd.read_csv('/content/train_ctrUa4K.csv')
StatCalc=statCalc()

M_mean=df.loc[df['Gender']=='Male','ApplicantIncome'].mean()
F_mean=df.loc[df['Gender']=='Female','ApplicantIncome'].mean()
M_std=df.loc[df['Gender']=='Male','ApplicantIncome'].std()
F_std=df.loc[df['Gender']=='Female','ApplicantIncome'].std()
no_of_M=df.loc[df['Gender']=='Male','ApplicantIncome'].count()
no_of_F=df.loc[df['Gender']=='Female','ApplicantIncome'].count()

z,p= StatCalc.Zstat2(M_mean,F_mean,0,M_std,F_std,no_of_M,no_of_F)
print(f'Z:{z},p: {p}')
```
